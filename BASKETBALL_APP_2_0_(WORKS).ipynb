```python
import pandas as pd

url = "https://www.basketball-reference.com/leagues/NBA_2025_per_game.html"

df = pd.read_html(url)[0]

print("Players:", len(df))
print(df.head())
```

```python
# Clean the NBA player data

# Remove repeated header rows
df = df[df["Player"] != "Player"].copy()

# Convert important stats to numbers
stats_to_clean = ["G", "PTS", "TRB", "AST", "STL", "BLK", "3P%", "FT%"]

for stat in stats_to_clean:
    df[stat] = pd.to_numeric(df[stat], errors="coerce")

# Remove players with missing data
df = df.dropna(subset=stats_to_clean)

print("Clean players:", len(df))
print(df[["Player", "PTS", "TRB", "AST", "STL", "BLK", "3P%", "FT%"]].head(10))
```

```python
# Cell 7 — create comparison dataframe

comparison_df = df[
    [
        "Player",
        "G",
        "PTS",
        "TRB",
        "AST",
        "STL",
        "BLK",
        "3P%",
        "FT%"
    ]
].copy()

# Rename columns to make the code easier to work with
comparison_df = comparison_df.rename(columns={
    "Player": "player_name",
    "G": "games",
    "PTS": "ppg",
    "TRB": "rpg",
    "AST": "apg",
    "STL": "spg",
    "BLK": "bpg",
    "3P%": "three_pct",
    "FT%": "ft_pct"
})

comparison_df = comparison_df.reset_index(drop=True)

print("Comparison dataframe:", comparison_df.shape)
print(comparison_df.head())
```

```python
# Cell 8 — add scoring efficiency

comparison_df["ppg_per_game"] = comparison_df["ppg"]

# Make sure percentage columns are numeric
comparison_df["three_pct"] = pd.to_numeric(
    comparison_df["three_pct"], errors="coerce"
)

comparison_df["ft_pct"] = pd.to_numeric(
    comparison_df["ft_pct"], errors="coerce"
)

# Remove players missing important stats
comparison_df = comparison_df.dropna(
    subset=[
        "ppg",
        "rpg",
        "apg",
        "spg",
        "bpg",
        "three_pct",
        "ft_pct"
    ]
).copy()

print("Players ready:", len(comparison_df))
print(comparison_df.head())
```

```python
# Cell 9 — stats used by the similarity model

comparison_stats = [
    "ppg",
    "rpg",
    "apg",
    "spg",
    "bpg",
    "three_pct",
    "ft_pct"
]

print("Stats used for comparison:")
for stat in comparison_stats:
    print("-", stat)
```

```python
# Cell 8 — clean and prepare the stats

stat_columns = [
    "games",
    "ppg",
    "rpg",
    "apg",
    "spg",
    "bpg",
    "three_pct",
    "ft_pct"
]

# Make sure all stat columns are numeric
for stat in stat_columns:
    comparison_df[stat] = pd.to_numeric(
        comparison_df[stat],
        errors="coerce"
    )

# Remove players missing too many stats
comparison_df = comparison_df.dropna(subset=stat_columns)

# Reset index
comparison_df = comparison_df.reset_index(drop=True)

print("Cleaned dataframe:", comparison_df.shape)
print()
print(comparison_df.head())
```

```python
# Cell 9 - player similarity function

def player_similarity(player1, player2):

    # Find both players
    p1 = comparison_df[
        comparison_df["player_name"].str.lower() == player1.lower()
    ]

    p2 = comparison_df[
        comparison_df["player_name"].str.lower() == player2.lower()
    ]

    # Make sure both players exist
    if p1.empty or p2.empty:
        print("One or both players were not found.")
        return None

    p1 = p1.iloc[0]
    p2 = p2.iloc[0]

    # How important each stat is
    weights = {
        "ppg": 0.20,
        "rpg": 0.15,
        "apg": 0.15,
        "spg": 0.10,
        "bpg": 0.10,
        "three_pct": 0.15,
        "ft_pct": 0.15
    }

    total_score = 0
    total_weight = 0

    # Compare each stat
    for stat, weight in weights.items():

        a = p1[stat]
        b = p2[stat]

        if pd.isna(a) or pd.isna(b):
            continue

        # Calculate how close the two stats are
        denominator = abs(a) + abs(b)

        if denominator == 0:
            similarity = 100
        else:
            difference = abs(a - b)
            similarity = 100 * (1 - difference / denominator)

        total_score += similarity * weight
        total_weight += weight

    if total_weight == 0:
        return None

    return round(total_score / total_weight, 2)
```

```python
# Cell 10 - test player similarity

test_pairs = [
    ("Nikola Jokic", "Joel Embiid"),
    ("Nikola Jokic", "Luka Doncic"),
    ("Stephen Curry", "Nikola Jokic"),
    ("LeBron James", "Rudy Gobert")
]

for p1, p2 in test_pairs:
    score = player_similarity(p1, p2)

    print(f"{p1} vs {p2}")

    if score is not None:
        print(f"Similarity: {score}%")
    else:
        print("Could not calculate similarity.")

    print("-" * 40)
```

```python
# Cell 11 - improve player name matching

import unicodedata

def normalize_name(name):
    name = str(name).lower().strip()

    # Remove accents
    name = unicodedata.normalize("NFKD", name)
    name = "".join(
        char for char in name
        if not unicodedata.combining(char)
    )

    return name


# Test the name normalization
print(normalize_name("Nikola Jokić"))
print(normalize_name("Nikola Jokic"))
print(normalize_name("Luka Dončić"))
print(normalize_name("Luka Doncic"))
```

```python
# Cell 12 - updated similarity function

def player_similarity(player1, player2):

    name1 = normalize_name(player1)
    name2 = normalize_name(player2)

    p1 = comparison_df[
        comparison_df["player_name"].apply(normalize_name) == name1
    ]

    p2 = comparison_df[
        comparison_df["player_name"].apply(normalize_name) == name2
    ]

    if p1.empty or p2.empty:
        print(f"Could not find: {player1} or {player2}")
        return None

    p1 = p1.iloc[0]
    p2 = p2.iloc[0]

    weights = {
        "ppg": 0.20,
        "rpg": 0.15,
        "apg": 0.15,
        "spg": 0.10,
        "bpg": 0.10,
        "three_pct": 0.15,
        "ft_pct": 0.15
    }

    total_score = 0
    total_weight = 0

    for stat, weight in weights.items():

        a = p1[stat]
        b = p2[stat]

        if pd.isna(a) or pd.isna(b):
            continue

        denominator = abs(a) + abs(b)

        if denominator == 0:
            similarity = 100
        else:
            difference = abs(a - b)
            similarity = 100 * (1 - difference / denominator)

        total_score += similarity * weight
        total_weight += weight

    if total_weight == 0:
        return None

    return round(total_score / total_weight, 2)
```

```python
# Cell 13 - find the most similar players

def find_similar_players(player_name, n=10):

    target_name = normalize_name(player_name)

    # Check that the player exists
    matches = comparison_df[
        comparison_df["player_name"].apply(normalize_name) == target_name
    ]

    if matches.empty:
        print(f"Player not found: {player_name}")
        return []

    results = []

    # Compare against every other player
    for _, player in comparison_df.iterrows():

        other_name = player["player_name"]

        # Don't compare the player to themselves
        if normalize_name(other_name) == target_name:
            continue

        score = player_similarity(player_name, other_name)

        if score is not None:
            results.append((other_name, score))

    # Sort highest similarity first
    results.sort(key=lambda x: x[1], reverse=True)

    print(f"Most similar players to {player_name}:")
    print()

    for name, score in results[:n]:
        print(f"{name}: {score}%")

    return results[:n]
```

```python
# Cell 16 - combine multiple rows for the same player

def combine_player_rows(group):
    total_games = group["games"].sum()

    result = {}

    # Player name comes from the group name
    result["player_name"] = group.name

    # Combine games
    result["games"] = total_games

    # Stats that are per-game: weight by games played
    per_game_stats = [
        "ppg",
        "rpg",
        "apg",
        "spg",
        "bpg"
    ]

    for stat in per_game_stats:
        result[stat] = (
            (group[stat] * group["games"]).sum()
            / total_games
        )

    # Shooting percentages: weight by games played
    result["three_pct"] = (
        (group["three_pct"] * group["games"]).sum()
        / total_games
    )

    result["ft_pct"] = (
        (group["ft_pct"] * group["games"]).sum()
        / total_games
    )

    return pd.Series(result)


# Combine duplicate player rows
comparison_df = (
    comparison_df
    .groupby("player_name")
    .apply(combine_player_rows, include_groups=False)
    .reset_index(drop=True)
)

print("Combined dataframe:", comparison_df.shape)
print()
print(comparison_df.head())
```

```python
# Cell 18 - calculate player role scores

# Create a copy so we don't mess up the main dataframe
roles_df = comparison_df.copy()

# Normalize stats from 0-1 so we can compare different statistics
stat_columns = [
    "ppg",
    "rpg",
    "apg",
    "spg",
    "bpg",
    "three_pct",
    "ft_pct"
]

for stat in stat_columns:
    min_val = roles_df[stat].min()
    max_val = roles_df[stat].max()

    if max_val > min_val:
        roles_df[stat + "_norm"] = (
            (roles_df[stat] - min_val) /
            (max_val - min_val)
        )
    else:
        roles_df[stat + "_norm"] = 0


# Role scores
roles_df["scoring_score"] = (
    roles_df["ppg_norm"] * 0.65 +
    roles_df["ft_pct_norm"] * 0.15 +
    roles_df["three_pct_norm"] * 0.20
)

roles_df["playmaking_score"] = (
    roles_df["apg_norm"] * 0.75 +
    roles_df["rpg_norm"] * 0.10 +
    roles_df["ppg_norm"] * 0.15
)

roles_df["rebounding_score"] = (
    roles_df["rpg_norm"] * 0.80 +
    roles_df["bpg_norm"] * 0.10 +
    roles_df["ppg_norm"] * 0.10
)

roles_df["defense_score"] = (
    roles_df["spg_norm"] * 0.45 +
    roles_df["bpg_norm"] * 0.40 +
    roles_df["rpg_norm"] * 0.15
)

roles_df["shooting_score"] = (
    roles_df["three_pct_norm"] * 0.70 +
    roles_df["ft_pct_norm"] * 0.30
)

roles_df["all_around_score"] = (
    roles_df["ppg_norm"] * 0.25 +
    roles_df["rpg_norm"] * 0.20 +
    roles_df["apg_norm"] * 0.20 +
    roles_df["spg_norm"] * 0.15 +
    roles_df["bpg_norm"] * 0.10 +
    roles_df["three_pct_norm"] * 0.10
)

print("Role scores created!")
print()
print(roles_df[
    [
        "player_name",
        "scoring_score",
        "playmaking_score",
        "rebounding_score",
        "defense_score",
        "shooting_score",
        "all_around_score"
    ]
].head(10))
```

```python
# Cell 19 - assign primary and optional secondary roles

role_columns = {
    "scoring_score": "Primary Scorer",
    "playmaking_score": "Playmaker",
    "rebounding_score": "Rebounder",
    "defense_score": "Defensive Specialist",
    "shooting_score": "Shooter",
    "all_around_score": "All-Around"
}

role_score_columns = list(role_columns.keys())

# Sort each player's role scores from highest to lowest
sorted_roles = roles_df[role_score_columns].apply(
    lambda row: row.sort_values(ascending=False).index.tolist(),
    axis=1
)

# Primary role
roles_df["primary_role"] = sorted_roles.apply(
    lambda x: role_columns[x[0]]
)

# Second-highest role
roles_df["secondary_role"] = sorted_roles.apply(
    lambda x: role_columns[x[1]]
)

# Get the top two scores
roles_df["primary_role_score"] = roles_df.apply(
    lambda row: row[sorted_roles.loc[row.name][0]],
    axis=1
)

roles_df["secondary_role_score"] = roles_df.apply(
    lambda row: row[sorted_roles.loc[row.name][1]],
    axis=1
)

# Only keep the secondary role if it is reasonably close
role_gap = (
    roles_df["primary_role_score"]
    - roles_df["secondary_role_score"]
)

roles_df.loc[role_gap > 0.20, "secondary_role"] = None

print("Roles assigned!")
print()

print(
    roles_df[
        [
            "player_name",
            "primary_role",
            "secondary_role",
            "primary_role_score",
            "secondary_role_score"
        ]
    ].head(20)
)
```

```python
# Cell 20 - role-aware player similarity

def role_aware_similarity(player1, player2):

    name1 = normalize_name(player1)
    name2 = normalize_name(player2)

    p1 = roles_df[
        roles_df["player_name"].apply(normalize_name) == name1
    ]

    p2 = roles_df[
        roles_df["player_name"].apply(normalize_name) == name2
    ]

    if p1.empty or p2.empty:
        print(f"Could not find: {player1} or {player2}")
        return None

    p1 = p1.iloc[0]
    p2 = p2.iloc[0]

    # -------------------------
    # 1. STAT SIMILARITY = 75%
    # -------------------------

    weights = {
        "ppg": 0.20,
        "rpg": 0.15,
        "apg": 0.15,
        "spg": 0.10,
        "bpg": 0.10,
        "three_pct": 0.15,
        "ft_pct": 0.15
    }

    stat_score = 0
    total_weight = 0

    for stat, weight in weights.items():

        a = p1[stat]
        b = p2[stat]

        if pd.isna(a) or pd.isna(b):
            continue

        denominator = abs(a) + abs(b)

        if denominator == 0:
            similarity = 100
        else:
            similarity = 100 * (
                1 - abs(a - b) / denominator
            )

        stat_score += similarity * weight
        total_weight += weight

    stat_score = stat_score / total_weight

    # -------------------------
    # 2. ROLE SIMILARITY = 25%
    # -------------------------

    role_scores = [
        "scoring_score",
        "playmaking_score",
        "rebounding_score",
        "defense_score",
        "shooting_score",
        "all_around_score"
    ]

    role_differences = []

    for role in role_scores:

        difference = abs(
            p1[role] - p2[role]
        )

        role_differences.append(difference)

    average_role_difference = np.mean(role_differences)

    role_score = 100 * (
        1 - average_role_difference
    )

    role_score = max(0, min(100, role_score))

    # -------------------------
    # 3. FINAL SCORE
    # -------------------------

    final_score = (
        stat_score * 0.75
        + role_score * 0.25
    )

    return round(final_score, 2)
```

```python
# Cell 21 - test role-aware similarity

test_pairs = [
    ("Nikola Jokic", "Joel Embiid"),
    ("Nikola Jokic", "Luka Doncic"),
    ("Stephen Curry", "Damian Lillard"),
    ("Jrue Holiday", "Marcus Smart")
]

for p1, p2 in test_pairs:

    score = role_aware_similarity(p1, p2)

    print(f"{p1} vs {p2}: {score}%")
```

```python
# Cell 22 - find similar players

def find_similar_players(player_name, n=10):

    if not any(
        roles_df["player_name"].apply(normalize_name)
        == normalize_name(player_name)
    ):
        print(f"Could not find {player_name}")
        return

    results = []

    for other_player in roles_df["player_name"]:

        if normalize_name(other_player) == normalize_name(player_name):
            continue

        score = role_aware_similarity(
            player_name,
            other_player
        )

        if score is not None:
            results.append(
                (other_player, score)
            )

    results.sort(
        key=lambda x: x[1],
        reverse=True
    )

    print(f"Most similar players to {player_name}:")
    print()

    for player, score in results[:n]:
        print(f"{player}: {score}%")

    return results[:n]
```

```python
# Cell 28 - flexible player lookup

import unicodedata

def normalize_name(name):
    """Normalize a name by removing accents and capitalization."""
    name = str(name).lower().strip()
    name = unicodedata.normalize("NFKD", name)
    name = "".join(
        char for char in name
        if not unicodedata.combining(char)
    )
    return name


def find_player(name):
    """
    Find a player using either their full name or last name.
    Examples:
        'Nikola Jokic' -> Nikola Jokić
        'Jokic' -> Nikola Jokić
        'jokic' -> Nikola Jokić
        'Doncic' -> Luka Dončić
    """

    if not isinstance(name, str) or not name.strip():
        return None

    search_name = normalize_name(name)

    # Normalize all player names
    normalized_names = comparison_df["player_name"].apply(normalize_name)

    # 1. Try exact full-name match
    matches = comparison_df[
        normalized_names == search_name
    ]

    if not matches.empty:
        return matches.iloc[0]

    # 2. Try last-name match
    last_name_matches = comparison_df[
        normalized_names.str.split().str[-1] == search_name
    ]

    if len(last_name_matches) == 1:
        return last_name_matches.iloc[0]

    # 3. Try partial name match
    partial_matches = comparison_df[
        normalized_names.str.contains(search_name, na=False)
    ]

    if len(partial_matches) == 1:
        return partial_matches.iloc[0]

    # Couldn't uniquely identify player
    return None
```

```python
# Cell 29 - clean player comparison report

def compare_players(player1, player2):
    score = role_aware_similarity(player1, player2)

    p1 = find_player(player1)
    p2 = find_player(player2)

    if p1 is None:
        print(f"Could not find {player1}.")
        return

    if p2 is None:
        print(f"Could not find {player2}.")
        return

    if score is None:
        print(f"Could not calculate similarity.")
        return

    print("=" * 55)
    print(f"{p1['player_name']} vs {p2['player_name']}")
    print("=" * 55)

    print(f"Similarity Score: {score:.2f}%")
    print()

    stats = [
        "games",
        "ppg",
        "rpg",
        "apg",
        "spg",
        "bpg",
        "three_pct",
        "ft_pct"
    ]

    print(f"{'Stat':<15} {'Player 1':>12} {'Player 2':>12}")
    print("-" * 42)

    for stat in stats:
        print(
            f"{stat:<15}"
            f"{p1[stat]:>12.2f}"
            f"{p2[stat]:>12.2f}"
        )

    print("=" * 55)
```

```python
# Cell 30 - improved similarity table

def get_similarity_table(player_name, n=10):
    """
    Get clean similarity results using flexible player lookup.
    """

    player = find_player(player_name)

    if player is None:
        print(f"Could not find {player_name}")
        return pd.DataFrame(columns=["player_name", "similarity"])

    # Use the actual database name
    actual_name = player["player_name"]

    results = get_similar_players(actual_name, n)

    if results is None or len(results) == 0:
        return pd.DataFrame(columns=["player_name", "similarity"])

    results = results.copy()
    results["similarity"] = results["similarity"].round(2)

    return results.reset_index(drop=True)
```

```python
# Cell 31 - user-friendly similarity search

def search_similar_players(player_name, n=10):
    """
    Find the most similar NBA players to a given player.
    """

    player = find_player(player_name)

    if player is None:
        print(f"❌ Could not find '{player_name}'.")
        print("Check the spelling and try again.")
        return None

    results = get_similarity_table(player["player_name"], n)

    print("=" * 55)
    print(f"Players similar to {player['player_name']}")
    print("=" * 55)
    print()

    if results.empty:
        print("No similar players found.")
        return results

    for i, row in results.iterrows():
        print(
            f"{i + 1:>2}. "
            f"{row['player_name']:<25} "
            f"{row['similarity']:.2f}%"
        )

    return results
```

```python
# Cell 34 - player name suggestions

def suggest_players(name, limit=5):
    """
    Suggest players based on a partial or misspelled name.
    """

    if not isinstance(name, str) or not name.strip():
        return []

    search_name = normalize_name(name)

    players = comparison_df["player_name"].unique()

    # Players containing the search text
    matches = [
        player for player in players
        if search_name in normalize_name(player)
    ]

    return matches[:limit]


# Test
print(suggest_players("Jok"))
print(suggest_players("Steph"))
print(suggest_players("Donc"))
```

```python
# Cell 36 - similarity results with player roles

def get_similarity_with_roles(player_name, n=10):

    results = get_similarity_table(player_name, n)

    if results is None or results.empty:
        return None

    role_info = roles_df[
        [
            "player_name",
            "primary_role",
            "secondary_role"
        ]
    ].copy()

    results = results.merge(
        role_info,
        on="player_name",
        how="left"
    )

    return results[
        [
            "player_name",
            "similarity",
            "primary_role",
            "secondary_role"
        ]
    ]


# Test
jokic_results = get_similarity_with_roles("Jokic", 10)

print(jokic_results.to_string(index=False))
```

```python
# Cell 37 - role-aware player comparison

import unicodedata

def normalize_name(name):
    """Remove accents and make names lowercase."""
    name = unicodedata.normalize("NFKD", str(name))
    name = "".join(
        char for char in name
        if not unicodedata.combining(char)
    )
    return name.lower().strip()


def resolve_player_name(name):
    """Convert user input like 'Jokic' or 'Luka' to the full player name."""

    search = normalize_name(name)

    # Normalize all player names
    normalized_names = comparison_df["player_name"].apply(normalize_name)

    # 1. Exact match
    matches = comparison_df[normalized_names == search]

    if not matches.empty:
        return matches.iloc[0]["player_name"]

    # 2. Last-name match
    last_name_matches = comparison_df[
        normalized_names.str.split().str[-1] == search
    ]

    if not last_name_matches.empty:
        return last_name_matches.iloc[0]["player_name"]

    # 3. First-name match
    first_name_matches = comparison_df[
        normalized_names.str.split().str[0] == search
    ]

    if not first_name_matches.empty:
        return first_name_matches.iloc[0]["player_name"]

    # 4. Partial match
    partial_matches = comparison_df[
        normalized_names.str.contains(search, na=False)
    ]

    if not partial_matches.empty:
        return partial_matches.iloc[0]["player_name"]

    return None


def compare_players_with_roles(player1, player2):

    # Resolve names
    resolved_p1 = resolve_player_name(player1)
    resolved_p2 = resolve_player_name(player2)

    if resolved_p1 is None:
        print(f"Could not find {player1}")
        return

    if resolved_p2 is None:
        print(f"Could not find {player2}")
        return

    # Get player rows
    p1 = comparison_df[
        comparison_df["player_name"] == resolved_p1
    ].iloc[0]

    p2 = comparison_df[
        comparison_df["player_name"] == resolved_p2
    ].iloc[0]

    # Calculate similarity using the existing system
    score = role_aware_similarity(resolved_p1, resolved_p2)

    if score is None:
        return

    print("=" * 65)
    print(f"{resolved_p1} vs {resolved_p2}")
    print("=" * 65)

    print(f"Similarity Score: {score:.2f}%")
    print()

    # Get roles
    p1_role = roles_df[
        roles_df["player_name"] == resolved_p1
    ]

    p2_role = roles_df[
        roles_df["player_name"] == resolved_p2
    ]

    print("Roles")
    print("-" * 65)

    if not p1_role.empty:
        print(
            f"{resolved_p1}: "
            f"{p1_role.iloc[0]['primary_role']} | "
            f"Secondary: {p1_role.iloc[0]['secondary_role']}"
        )

    if not p2_role.empty:
        print(
            f"{resolved_p2}: "
            f"{p2_role.iloc[0]['primary_role']} | "
            f"Secondary: {p2_role.iloc[0]['secondary_role']}"
        )

    print()

    print(f"{'Stat':<15}{'Player 1':>15}{'Player 2':>15}")
    print("-" * 45)

    stats = [
        "games",
        "ppg",
        "rpg",
        "apg",
        "spg",
        "bpg",
        "three_pct",
        "ft_pct"
    ]

    for stat in stats:
        print(
            f"{stat:<15}"
            f"{p1[stat]:>15.2f}"
            f"{p2[stat]:>15.2f}"
        )

    print("=" * 65)
```

```python
# Cell 41 - accent-insensitive player search

import unicodedata

def normalize_name(name):
    return ''.join(
        c for c in unicodedata.normalize("NFD", str(name))
        if unicodedata.category(c) != "Mn"
    ).lower()


def find_player(search_term):
    search_term = search_term.strip()

    # Exact match
    exact = comparison_df[
        comparison_df["player_name"].str.lower() == search_term.lower()
    ]

    if len(exact) == 1:
        return exact.iloc[0].to_dict()

    # Otherwise get suggestions
    suggestions = find_player_suggestions(search_term)

    if len(suggestions) == 1:
        player = comparison_df[
            comparison_df["player_name"] == suggestions[0]
        ]

        if len(player) == 1:
            return player.iloc[0].to_dict()

    # Multiple matches → return the names
    if len(suggestions) > 1:
        return suggestions

    return None
```

```python
def get_similarity_breakdown(player1, player2):
    p1 = find_player(player1)
    p2 = find_player(player2)

    if p1 is None or p2 is None:
        print("Could not find one or both players.")
        return None

    if isinstance(p1, list) or isinstance(p2, list):
        print("Multiple players found. Please use a more specific name.")
        return None

    stats = [
        "ppg",
        "rpg",
        "apg",
        "spg",
        "bpg",
        "three_pct",
        "ft_pct"
    ]

    breakdown = {}

    for stat in stats:
        v1 = p1[stat]
        v2 = p2[stat]

        # Similarity based on how close the two values are
        if max(abs(v1), abs(v2)) == 0:
            similarity = 100
        else:
            similarity = (
                1 - abs(v1 - v2) / max(abs(v1), abs(v2))
            ) * 100

        breakdown[stat] = max(0, similarity)

    return breakdown
```

```python
def print_similarity_breakdown(player1, player2):
    p1 = find_player(player1)
    p2 = find_player(player2)

    if p1 is None or p2 is None:
        print("Could not find one or both players.")
        return

    if isinstance(p1, list) or isinstance(p2, list):
        print("Multiple players found. Please use a more specific name.")
        return

    breakdown = get_similarity_breakdown(player1, player2)

    if breakdown is None:
        return

    overall = sum(breakdown.values()) / len(breakdown)

    print("=" * 60)
    print(f"{p1['player_name']} vs {p2['player_name']}")
    print(f"Overall Similarity: {overall:.2f}%")
    print("=" * 60)

    print()
    print(f"{'Stat':<12}{p1['player_name']:<14}{p2['player_name']:<14}{'Similarity'}")
    print("-" * 60)

    stat_names = {
        "ppg": "PPG",
        "rpg": "RPG",
        "apg": "APG",
        "spg": "SPG",
        "bpg": "BPG",
        "three_pct": "3P%",
        "ft_pct": "FT%"
    }

    for stat, score in breakdown.items():
        name = stat_names[stat]

        if stat in ["three_pct", "ft_pct"]:
            v1 = f"{p1[stat] * 100:.2f}%"
            v2 = f"{p2[stat] * 100:.2f}%"
        else:
            v1 = f"{p1[stat]:.2f}"
            v2 = f"{p2[stat]:.2f}"

        print(f"{name:<12}{v1:<14}{v2:<14}{score:.2f}%")
```
